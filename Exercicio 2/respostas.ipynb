<h3><strong>1- Filtros de Suavização Linear</strong></h3>
<h4><strong>1.1- Comparação entre Filtros de Suavização</strong></h4>

Escolha uma imagem com diferentes tipos de ruído (por exemplo, sal e pimenta e ruído gaussiano).
Aplique os seguintes filtros de suavizac¸ao: média, mediana e média ponderada. Compare os resultados e discuta qual filtro e mais eficaz para cada tipo de ruído.


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Carregar a imagem com ruído Sal e Pimenta
imagem = cv2.imread('img\\mirrage-saltpepper.tiff', cv2.IMREAD_GRAYSCALE)

# Aplicação dos filtros na Imagem
filtro_media = cv2.blur(imagem, (3, 3))  # Filtro de média
filtro_mediana = cv2.medianBlur(imagem, 3)  # Filtro de mediana
# Definir um kernel para média ponderada
kernel_ponderado = np.array([[1, 2, 1], [2, 4, 2], [1, 2, 1]]) / 16
filtro_media_ponderada = cv2.filter2D(imagem, -1, kernel_ponderado)

# Exibição dos resultados
plt.figure(figsize=(12, 8))
plt.subplot(1, 4, 1), plt.imshow(imagem, cmap='gray'), plt.title('Original')
plt.subplot(1, 4, 2), plt.imshow(filtro_media, cmap='gray'), plt.title('Filtro de Média')
plt.subplot(1, 4, 3), plt.imshow(filtro_mediana, cmap='gray'), plt.title('Filtro de Mediana')
plt.subplot(1, 4, 4), plt.imshow(filtro_media_ponderada, cmap='gray'), plt.title('Filtro de Média Ponderada')
plt.show()




# Carregar a imagem com ruído gaussiano
imagem = cv2.imread('img\gaussiano.tiff', cv2.IMREAD_GRAYSCALE)

# Aplicação dos filtros na imagem com ruido gaussiano
filtro_media = cv2.blur(imagem, (3, 3))  # Filtro de média
filtro_mediana = cv2.medianBlur(imagem, 3)  # Filtro de mediana
# Definir um kernel para média ponderada
kernel_ponderado = np.array([[1, 2, 1], [2, 4, 2], [1, 2, 1]]) / 16
filtro_media_ponderada = cv2.filter2D(imagem, -1, kernel_ponderado)

# Exibição dos resultados
plt.figure(figsize=(12, 8))
plt.subplot(1, 4, 1), plt.imshow(imagem, cmap='gray'), plt.title('Original')
plt.subplot(1, 4, 2), plt.imshow(filtro_media, cmap='gray'), plt.title('Filtro de Média')
plt.subplot(1, 4, 3), plt.imshow(filtro_mediana, cmap='gray'), plt.title('Filtro de Mediana')
plt.subplot(1, 4, 4), plt.imshow(filtro_media_ponderada, cmap='gray'), plt.title('Filtro de Média Ponderada')
plt.show()
```

<strong>Análise dos resultados: </strong> Percebe-se que a o Filtro de Mediana é mais eficaz para ruídos do tipo "sal e pimenta". Já para ruídos gaussianos, nota-se que a melhor opção é o Filtro de Média Ponderada. 

<h4><strong>1.2- Filtro da Média Ponderada</strong></h4>

Implemente um filtro da media ponderada com diferentes configurações de pesos. Aplique este filtro em uma imagem e compare os resultados com o filtro de media simples. Discuta como a variação dos pesos afeta a suavizac¸ao da imagem. 

```python
# Carregar a imagem
imagem = cv2.imread('img\\tank-gaussiano.tiff', cv2.IMREAD_GRAYSCALE)

# Definir diferentes kernels ponderados
kernel_1 = np.array([[1, 1, 1], [1, 1, 1], [1, 1, 1]]) / 9  # Filtro de Média Simples
kernel_2 = np.array([[5, 3, 6], [5, 4, 2], [7, 1, 1]]) / 16  # Filtro de Média Ponderada
kernel_3 = np.array([[1, 2, 1], [2, 4, 2], [1, 2, 1]]) / 16  # Filtro de Média Ponderada

# Aplicar os filtros
filtro_media_simples = cv2.filter2D(imagem, -1, kernel_1)
filtro_media_ponderada = cv2.filter2D(imagem, -1, kernel_2)
filtro_media_ponderada2 = cv2.filter2D(imagem, -1, kernel_3)

# Exibição dos resultados
plt.figure(figsize=(16, 12))
plt.subplot(1, 4, 1), plt.imshow(imagem, cmap='gray'), plt.title('Original')
plt.subplot(1, 4, 2), plt.imshow(filtro_media_simples, cmap='gray'), plt.title('Média Simples')
plt.subplot(1, 4, 3), plt.imshow(filtro_media_ponderada, cmap='gray'), plt.title('Média Ponderada')
plt.subplot(1, 4, 4), plt.imshow(filtro_media_ponderada2, cmap='gray'), plt.title('Média Ponderada 2')
plt.show()
```

<strong>Análise dos resultados:</strong> A Média Ponderada permite ajustar a suavização da imagem de acordo com o peso dado aos pixels centrais. Maior peso nos pixels centrais preserva mais os detalhes, enquanto pesos uniformes resultam em uma suavização mais agressiva.

<h3><strong>2- Filtros de Detecção de Bordas</strong></h3>
<h4><strong>2.1- Comparação entre Filtros de Detecção de Bordas</strong></h4>

Aplique os seguintes filtros de deteção de bordas em uma imagem: <i>Roberts</i>, <i>Prewitt</i>, <i>Sobel</i> e <i>Laplaciano</i>. Compare os resultados em termos de nitidez das bordas detectadas e discuta qual filtro seria mais adequado para diferentes tipos de imagens e cenarios.


```python
# Carregar a primeira imagem
imagem = cv2.imread('img\\casa.tiff', cv2.IMREAD_GRAYSCALE)

# Aplicação dos filtros de borda na primeira imagem
filtro_roberts = cv2.filter2D(imagem, -1, np.array([[1, 0], [0, -1]])) + cv2.filter2D(imagem, -1, np.array([[0, 1], [-1, 0]]))
filtro_prewitt = cv2.filter2D(imagem, -1, np.array([[-1, 0, 1], [-1, 0, 1], [-1, 0, 1]])) + cv2.filter2D(imagem, -1, np.array([[1, 1, 1], [0, 0, 0], [-1, -1, -1]]))
filtro_sobel = cv2.Sobel(imagem, cv2.CV_64F, 1, 0, ksize=3) + cv2.Sobel(imagem, cv2.CV_64F, 0, 1, ksize=3)
filtro_laplaciano = cv2.Laplacian(imagem, cv2.CV_64F)

# Exibição dos resultados
plt.figure(figsize=(16, 12))
plt.subplot(1, 5, 1), plt.imshow(imagem, cmap='gray'), plt.title('Original')
plt.subplot(1, 5, 2), plt.imshow(filtro_roberts, cmap='gray'), plt.title('Filtro Roberts')
plt.subplot(1, 5, 3), plt.imshow(filtro_prewitt, cmap='gray'), plt.title('Filtro Prewitt')
plt.subplot(1, 5, 4), plt.imshow(filtro_sobel, cmap='gray'), plt.title('Filtro Sobel')
plt.subplot(1, 5, 5), plt.imshow(filtro_laplaciano, cmap='gray'), plt.title('Filtro Laplaciano')
plt.show()
```

<strong>Análise dos resultados:</strong> Para a imagem, o filtro Prewitt gerou o melhor resultado (na minha opinião) dentre os filtros aplicados, um detalhe interessante é que esse filtro conseguiu definir bem a borda interna da chaminé assim como a textura. O filtro de Sobel é geralmente mais suave e melhor para bordas em imagens com ruído, enquanto o filtro de Roberts é mais sensível a bordas finas. O filtro Laplaciano pode destacar bordas mais grossas e é mais sensível a ruídos, enquanto o filtro de Prewitt é semelhante ao Sobel, mas menos sensível a bordas diagonais.

<h4><strong>2.3- Efeito da Intensidade do Filtro</strong></h4>

Aplique o filtro <i>Laplaciano</i> com diferentes intensidades (ajustando o <i>kernel</i>). Compare os resultados em termos de destaque das bordas e discuta como a intensidade do filtro pode influenciar a detecção de bordas em imagens com diferentes níveis de detalhe.

```python

# Função para aplicar o filtro Laplaciano com diferentes kernels
def aplicar_filtro_laplaciano(imagem, kernel):
    return cv2.filter2D(imagem, -1, kernel)

# Definir diferentes kernels do filtro Laplaciano
kernel_laplaciano_1 = np.array([[0, -1, 0],
                                [-1, 4, -1],
                                [0, -1, 0]], dtype=np.float32)

kernel_laplaciano_2 = np.array([[0, 0, -1, 0, 0],
                                [0, -1, -2, -1, 0],
                                [-1, -2, 16, -2, -1],
                                [0, -1, -2, -1, 0],
                                [0, 0, -1, 0, 0]], dtype=np.float32)

kernel_laplaciano_3 = np.array([[-1, -1, -1],
                                [-1,  8, -1],
                                [-1, -1, -1]], dtype=np.float32)

# Carregar a imagem em escala de cinza
#imagem = cv2.imread('img\\aviao.tiff', cv2.IMREAD_GRAYSCALE)
imagem = cv2.imread('img\\tank.tiff', cv2.IMREAD_GRAYSCALE)

# Aplicar o filtro Laplaciano com diferentes kernels
imagem_laplaciana_1 = aplicar_filtro_laplaciano(imagem, kernel_laplaciano_1)
imagem_laplaciana_2 = aplicar_filtro_laplaciano(imagem, kernel_laplaciano_2)
imagem_laplaciana_3 = aplicar_filtro_laplaciano(imagem, kernel_laplaciano_3)

# Exibir a imagem original e as imagens com diferentes filtros Laplacianos
plt.figure(figsize=(15, 10))
plt.subplot(2, 2, 1), plt.imshow(imagem, cmap='gray'), plt.title('Imagem Original')
plt.subplot(2, 2, 2), plt.imshow(imagem_laplaciana_1, cmap='gray'), plt.title('Filtro Laplaciano 1')
plt.subplot(2, 2, 3), plt.imshow(imagem_laplaciana_2, cmap='gray'), plt.title('Filtro Laplaciano 2')
plt.subplot(2, 2, 4), plt.imshow(imagem_laplaciana_3, cmap='gray'), plt.title('Filtro Laplaciano 3')
plt.show()
```

<strong>Análise dos resultados:</strong> <p>O filtro com intensidade maior (como o Kernel Laplaciano 2) tende a gerar bordas mais pronunciadas e evidentes, destacando mais os detalhes da imagem. No entanto, também pode aumentar o ruído.</p>
<p>Para imagens com mais detalhes e bordas suaves, um kernel menos intenso (como o Kernel Laplaciano 1 ou 3) pode fornecer uma detecção de borda mais equilibrada, evitando o exagero dos detalhes menores.</p>
